```python
catalog_name = "workspace"
schema_name  = "bronze"
expected_tables = ["FactSalesOrderDetails","FactSalesTerritoryDetails","FactSales",
                   "DimReseller","DimDate","DimProduct","DimCustomer"]
```

Listing the tables

```python
tables_in_bronze = spark.sql(f"SHOW TABLES IN {catalog_name}.{schema_name}")
display(tables_in_bronze)
```

```python
from pyspark.sql import Row

rows = []
for r in tables_in_bronze.collect():
    cnt = spark.table(f"{catalog_name}.{schema_name}.{r.tableName}").count()
    rows.append(Row(table_name=r.tableName, row_count=cnt))

row_counts_df = spark.createDataFrame(rows).orderBy("table_name")
display(row_counts_df)
```
